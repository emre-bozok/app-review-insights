# App Review Insights

A reproducible walkthrough using **synthetic reviews and pre-labelled fixtures**. This demo makes no API calls. Run from the repository root or the `notebooks/` directory. Install the root requirements and a Jupyter kernel first.


```python
from pathlib import Path
import json
import sys

ROOT = Path.cwd()
if not (ROOT / "review_insights.py").exists():
    ROOT = ROOT.parent
if not (ROOT / "review_insights.py").exists():
    raise RuntimeError("Start Jupyter in the repository root or notebooks directory.")
sys.path.insert(0, str(ROOT))
from review_insights import match_labels, summarize
```

## 1. Load synthetic fixtures

The labels below are manual examples. They demonstrate the reporting pipeline, not AI accuracy.


```python
reviews = json.loads((ROOT / "examples/reviews.synthetic.json").read_text())
labels = json.loads((ROOT / "examples/labels.synthetic.json").read_text())
print(f"Synthetic reviews: {len(reviews)}")
```

## 2. Validate and join by ID

Result order can differ from input order. Every review ID must occur exactly once.


```python
classified = match_labels(reviews, list(reversed(labels)))
report = summarize(classified)
print(json.dumps(report, ensure_ascii=False, indent=2))
```

## 3. Inspect category totals


```python
for item in report["categories"]:
    print(f'{item["category"]}: {item["count"]} ({item["percentage"]}%)')
```
